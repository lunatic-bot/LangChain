## Models integration with openAI, Google gemini and groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os


model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    api_key=os.getenv("GEMINI_API_KEY"),
)

model

```

```python
response = model.invoke("Hello, how are you?")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model


model = init_chat_model("google_genai:gemini-2.5-flash", api_key=os.getenv("GEMINI_API_KEY"))
response = model.invoke("Hello, how are you?")
print(response.content)
```

# Streaming
The model sends output token-by-token (or chunk-by-chunk) as it generates.


```python
# model.invoke
for chunk in model.stream("Write me a 200 word paragraph about Artificial Intelligence"):
    print(chunk.text, end="", flush=True)
```

## Batch
provides an efficient way to process multiple inputs simultaneously, significantly improving throughput for IO-bound tasks compared to sequential execution. It enables parallel processing across chains, models, and retrievers, with automatic handling of parallel execution via thread pools, commonly used to avoid RateLimitError

```python
responses = model.batch([
"Why do parrots talk?",
"How do airplanes fly?",
"What is the capital of the United States?"],
config={"max_concurrency":5} #limit to 5 parallel calls
)

for response in responses:
    print(response)
```
